```python
# Example of a nested loop
for i in range(1, 4):  # Outer loop
    for j in range(1, 4):  # Inner loop
        print(f"({i}, {j})", end=" ")
    print()  # New line after each row   
```

```python
num = int(input("Enter a num: "))
og = num
sum = 0
while num > 0:
    digi = num % 10
    sum = sum + digi ** 3
    num = num // 10

if sum == og:
    print("Armstrong number")
else:
    print(sum)
```

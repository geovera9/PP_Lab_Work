```python
# 1. Fibonacci Series*
n = int(input("Enter number of terms: "))
a, b = 0, 1

for i in range(n):
    print(a, end=" ")
    a, b = b, a + b
```

```python
# 2. Factorial using a while loop*
num = int(input("Enter a number: "))
fact = 1
i = 1

while i <= num:
    fact *= i
    i += 1

print(f"Factorial of {num} is {fact}")
```

```python
# 3. Prime Number Checking*
num = int(input("Enter a number: "))

if num <= 1:
    print("Not Prime")
else:
    is_prime = True
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    
    print("Prime" if is_prime else "Not Prime")
```

```python
# 4. Reverse Number and Digit Sum*
num = int(input("Enter a number: "))
original = num
reverse = 0
digit_sum = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    digit_sum += digit
    num //= 10

print(f"Original: {original}")
print(f"Reversed: {reverse}")
print(f"Sum of digits: {digit_sum}")
```

```python
# 5. Armstrong Number - Python Program*
num = int(input("Enter a number: "))
original = num
n = len(str(num))
sum_of_powers = 0

temp = num
while temp > 0:
    digit = temp % 10
    sum_of_powers += digit ** n
    temp //= 10

if sum_of_powers == original:
    print(f"{original} is an Armstrong Number")
else:
    print(f"{original} is not an Armstrong Number")
```
